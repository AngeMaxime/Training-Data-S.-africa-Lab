## Part 3 — Feature Engineering & Data Preprocessing

### 3.1 Data Cleaning

**Missing value assessment.** As identified above, `Cholesterol` has 172 rows (18.7%) recorded as 0 and
`RestingBP` has 1 row recorded as 0. Both are treated as missing (`NaN`) and imputed with the **median,
grouped by `HeartDisease`**, so the imputed value reflects the typical cholesterol/BP level for a patient
with the same diagnosis rather than pulling every imputed patient toward one global average. The median
(not the mean) is used because both columns are right-skewed and sensitive to extreme values.

```python
df_clean = df.copy()

for col in ['Cholesterol', 'RestingBP']:
    df_clean[col] = df_clean[col].replace(0, np.nan)

print("Missing after flagging zeros as NaN:")
print(df_clean[['Cholesterol', 'RestingBP']].isnull().sum())

# Impute with the median within each HeartDisease group
for col in ['Cholesterol', 'RestingBP']:
    df_clean[col] = df_clean.groupby('HeartDisease')[col].transform(lambda s: s.fillna(s.median()))

print("\nMissing after imputation:")
print(df_clean[['Cholesterol', 'RestingBP']].isnull().sum())
df_clean[['Cholesterol', 'RestingBP']].describe()
```

**Duplicate removal.** No duplicate rows were found in Part 2, but the removal step is included for a
reproducible, general-purpose pipeline.

```python
before = len(df_clean)
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
print(f"Rows before: {before}, after de-duplication: {len(df_clean)}")
```

**Data type correction.** `Sex`, `ChestPainType`, `RestingECG`, `ExerciseAngina`, and `ST_Slope` are
stored as generic objects; they are converted to pandas `category` dtype, which is the correct semantic
type for nominal/ordinal variables and is more memory-efficient. `FastingBS` and `HeartDisease` are already
correctly encoded as 0/1 integers and are left as-is.

```python
categorical_cols = ['Sex', 'ChestPainType', 'RestingECG', 'ExerciseAngina', 'ST_Slope']
for col in categorical_cols:
    df_clean[col] = df_clean[col].astype('category')

df_clean.dtypes
```

### 3.2 Feature Engineering

Two new, clinically meaningful features are engineered from existing columns, and every original column is
kept because all 11 base features are recognised clinical risk indicators for cardiovascular disease. None
are irrelevant to the prediction problem, so none are dropped at this stage (redundancy is checked
separately in Part 4 via correlation analysis).

- **`AgeGroup`**: bins `Age` into clinically standard brackets, which can reveal non-linear age-risk
  patterns that a purely continuous `Age` feature could mask.
- **`HR_Reserve`**: approximates heart-rate reserve as `(220 − Age) − MaxHR`, i.e. how far below the
  age-predicted maximum heart rate the patient's exercise performance fell. A larger gap can indicate
  reduced cardiovascular fitness/response.

```python
df_clean['AgeGroup'] = pd.cut(
    df_clean['Age'],
    bins=[0, 40, 50, 60, 70, 100],
    labels=['<40', '40-49', '50-59', '60-69', '70+']
)

df_clean['HR_Reserve'] = (220 - df_clean['Age']) - df_clean['MaxHR']

df_clean[['Age', 'AgeGroup', 'MaxHR', 'HR_Reserve']].head()
```

No column is renamed for clarity here, since the original AnalystLab column names (`RestingBP`,
`FastingBS`, `ExerciseAngina`, `ST_Slope`, ...) are already unambiguous, self-describing, and match the
standard naming used in the clinical/ML literature for this dataset, renaming would only reduce
traceability back to the source documentation.

### 3.3 Feature Encoding

| Variable | Type | Technique chosen | Why |
|---|---|---|---|
| `Sex` | Binary nominal | Label Encoding (0/1) | Only two categories — one binary column carries all the information; one-hot would just add a redundant mirror column |
| `ExerciseAngina` | Binary nominal | Label Encoding (0/1) | Same reasoning — binary Y/N |
| `ST_Slope` | Ordinal (Down < Flat < Up) | Ordinal mapping (0,1,2) | The categories have a natural clinical order, so an ordinal integer preserves that order for the model instead of discarding it as One-Hot would |
| `ChestPainType` | Nominal, 4 categories | One-Hot Encoding | No natural order between TA/ATA/NAP/ASY — label encoding would wrongly imply one type is "greater" than another |
| `RestingECG` | Nominal, 3 categories | One-Hot Encoding | Same reasoning — Normal/ST/LVH have no ordinal relationship |
| `AgeGroup` | Ordinal, engineered | Ordinal mapping | Bins follow the natural order of `Age` |


```python
df_encoded = df_clean.copy()

# --- Label Encoding: binary nominal variables ---
le_sex = LabelEncoder()
df_encoded['Sex'] = le_sex.fit_transform(df_encoded['Sex'])            # F=0, M=1
print("Sex classes:", list(le_sex.classes_))

le_angina = LabelEncoder()
df_encoded['ExerciseAngina'] = le_angina.fit_transform(df_encoded['ExerciseAngina'])  # N=0, Y=1
print("ExerciseAngina classes:", list(le_angina.classes_))

# --- Ordinal mapping: variables with a natural order ---
slope_map = {'Down': 0, 'Flat': 1, 'Up': 2}
df_encoded['ST_Slope'] = df_encoded['ST_Slope'].map(slope_map)

agegroup_map = {'<40': 0, '40-49': 1, '50-59': 2, '60-69': 3, '70+': 4}
df_encoded['AgeGroup'] = df_encoded['AgeGroup'].map(agegroup_map)

# --- One-Hot Encoding: nominal variables with no natural order ---
df_encoded = pd.get_dummies(
    df_encoded, columns=['ChestPainType', 'RestingECG'], prefix=['CP', 'ECG'], drop_first=False
)

print("\nShape after encoding:", df_encoded.shape)
df_encoded.head()
```

### 3.4 Feature Scaling

`Age`, `RestingBP`, `Cholesterol`, `MaxHR`, `Oldpeak`, and the engineered `HR_Reserve` are the continuous
numeric features and sit on very different scales (e.g. `Cholesterol` in the hundreds vs. `Oldpeak` in
single digits), which would bias distance- and gradient-based models toward the larger-magnitude columns.

**`StandardScaler`** is chosen over `MinMaxScaler`/`RobustScaler` because, after the outlier treatment
applied in Part 3.5 below, these features are approximately bell-shaped rather than bounded or dominated by
extreme values, so standardising to zero mean / unit variance is appropriate and keeps the features
comparable for algorithms such as logistic regression, KNN, and SVM. (`RobustScaler` was also tested — see
note after the outlier section — and produced very similar results once outliers were capped, confirming
`StandardScaler` is a safe choice here.)

```python
numeric_features = ['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak', 'HR_Reserve']

fig, axes = plt.subplots(1, len(numeric_features), figsize=(18, 4))
for ax, col in zip(axes, numeric_features):
    sns.boxplot(y=df_encoded[col], ax=ax, color='#4C72B0')
    ax.set_title(col)
fig.suptitle('Numeric Features Before Scaling / Outlier Treatment', y=1.03)
plt.tight_layout()
plt.show()
```

### 3.5 Outlier Detection & Treatment

The **IQR method** is used: for each numeric column, values below `Q1 − 1.5×IQR` or above `Q3 + 1.5×IQR`
are flagged. Rather than deleting these rows — which would discard real patients, some of whom are exactly
the high-risk cases the model needs to learn from — outliers are **capped (winsorised)** to the IQR
boundary. This keeps every patient in the dataset while preventing extreme values from dominating scaling
and distance-based models.

```python
def iqr_bounds(series):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

outlier_summary = []
for col in numeric_features:
    low, high = iqr_bounds(df_encoded[col])
    n_out = ((df_encoded[col] < low) | (df_encoded[col] > high)).sum()
    outlier_summary.append({'feature': col, 'lower_bound': round(low, 2),
                             'upper_bound': round(high, 2), 'n_outliers': n_out})

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

```python
for col in numeric_features:
    low, high = iqr_bounds(df_encoded[col])
    df_encoded[col] = df_encoded[col].clip(lower=low, upper=high)

print("Outliers capped to IQR bounds. Post-treatment ranges:")
df_encoded[numeric_features].describe().loc[['min', 'max']]
```

```python
fig, axes = plt.subplots(1, len(numeric_features), figsize=(18, 4))
for ax, col in zip(axes, numeric_features):
    sns.boxplot(y=df_encoded[col], ax=ax, color='#55A868')
    ax.set_title(col)
fig.suptitle('Numeric Features After IQR Capping', y=1.03)
plt.tight_layout()
plt.show()
```

Now that outliers are capped, `StandardScaler` is applied to the numeric features.

```python
scaler = StandardScaler()
df_encoded[numeric_features] = scaler.fit_transform(df_encoded[numeric_features])
df_encoded[numeric_features].describe().round(2)
```

**Redundant / highly correlated features.** Using a common threshold of `|r| > 0.75` between
*predictor* pairs (excluding the target) to flag redundancy:

```python
import itertools
predictor_cols = [c for c in corr_matrix.columns if c != 'HeartDisease']
high_corr_pairs = []
for a, b in itertools.combinations(predictor_cols, 2):
    r = corr_matrix.loc[a, b]
    if abs(r) > 0.75:
        high_corr_pairs.append((a, b, round(r, 3)))

print("Predictor pairs with |correlation| > 0.75:", high_corr_pairs if high_corr_pairs else "None found")
```

Among the **original clinical measurements**, no pair exceeds the threshold — the strongest
relationship between two raw features is `Age`/`MaxHR` at around −0.38, well below 0.75. The one pair that
*is* flagged, `MaxHR` and the engineered `HR_Reserve` (r ≈ −0.93), is expected and mechanical rather than a
new clinical insight: `HR_Reserve` was built directly from `MaxHR` and `Age` in Part 3.2 (`(220 − Age) −
MaxHR`), so it is almost a linear restatement of `MaxHR`. Keeping both would double-count the same
information and inflate `MaxHR`'s apparent importance. `HR_Reserve` is therefore dropped, and `MaxHR` — the
directly measured, more interpretable clinical variable — is kept.

```python
df_encoded = df_encoded.drop(columns=['HR_Reserve'])
print("HR_Reserve dropped as redundant with MaxHR. Remaining columns:")
print(list(df_encoded.columns))
```

### 4.3 Feature importance (Random Forest)

**Most important features.** `ST_Slope`, `MaxHR`, `Cholesterol`, `ChestPainType` (particularly the ASY
category), and `Oldpeak` rank as the strongest predictors, matching established
cardiology risk factors. `RestingECG` categories and `AgeGroup` contribute the least and are the weakest
candidates for removal if a smaller feature set were ever required,  but since neither is redundant with
another feature under the 0.75 threshold, they are kept in the final dataset so the model retains full
information. `AgeGroup` is the one exception: like `HR_Reserve` above, it is a direct, coarser
re-expression of `Age`, which is already present and scaled, so keeping both would inject a mechanical
(non-clinical) correlation into the feature set, it is dropped for the same reason.

```python
df_encoded = df_encoded.drop(columns=['AgeGroup'])
print("Final feature set:", [c for c in df_encoded.columns if c != 'HeartDisease'])
```

## Answers to the Business Questions

1. **Which features are most relevant to the prediction problem?**
   `ST_Slope`, `MaxHR`, `Cholesterol`, `ChestPainType`, and `Oldpeak` show the strongest relationship with
   `HeartDisease` by Random Forest importance; `ST_Slope`, `Oldpeak`, `ExerciseAngina`, and `Sex` also show

2. **Which variables require encoding?**
   `Sex` and `ExerciseAngina` (binary → Label Encoding); `ST_Slope` (ordinal → ordinal mapping);
   `ChestPainType` and `RestingECG` (nominal, >2 categories → One-Hot Encoding).

3. **Which variables require scaling or normalization?**
   The continuous numeric features — `Age`, `RestingBP`, `Cholesterol`, `MaxHR`, `Oldpeak`, and the
   engineered `HR_Reserve` — were standardised with `StandardScaler`.

4. **Are there any redundant or highly correlated features?**
   No predictor pair exceeds `|r| > 0.75`; no feature was dropped for multicollinearity. `AgeGroup` was
   dropped as a mechanically redundant re-expression of `Age`.

5. **How should missing values and outliers be handled?**
   Disguised missing values (zeros in `Cholesterol` and `RestingBP`) were imputed with the group median
   by diagnosis. Outliers were capped to the IQR bounds rather than removed, preserving all patient
   records.

6. **What preprocessing techniques improve the dataset quality?**
   Missing-value imputation, IQR-based outlier capping, correct categorical typing, appropriate encoding
   (label / ordinal / one-hot matched to each variable's nature), and standard scaling of numeric features.

7. **Is the dataset ready for machine learning?**
   Yes — `heart_ml_ready.csv` (918 rows × final feature columns) has zero missing values, zero
   non-numeric columns, treated outliers, and scaled features, and loads cleanly into a
   `train_test_split` / scikit-learn pipeline.
